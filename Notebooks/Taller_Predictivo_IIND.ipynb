```python
# PASO 2: Generación de Datos de Operaciones Logísticas
import pandas as pd
import numpy as np

np.random.seed(42)
n_pedidos = 1000

data = pd.DataFrame({
    'distancia_km': np.random.uniform(5, 50, n_pedidos),
    'volumen_m3': np.random.uniform(1, 10, n_pedidos),
    'trafico_index': np.random.uniform(0, 1, n_pedidos), # 1 es tráfico muy pesado
    'clima_index': np.random.uniform(0, 1, n_pedidos),   # 1 es clima adverso/tormenta
    'hora_pico': np.random.choice([0, 1], n_pedidos),
    'experiencia_conductor': np.random.randint(1, 15, n_pedidos),
    'retraso': np.random.choice([0, 1], n_pedidos) # El modelo aprenderá a predecir esto
})

# Lógica para que los datos tengan patrones reales que la IA pueda detectar:
data.loc[(data['trafico_index'] > 0.7) & (data['hora_pico'] == 1), 'retraso'] = 1
data.loc[data['distancia_km'] < 10, 'retraso'] = 0

print("✅ Base de datos logística cargada con éxito. Visualizando primeras filas:")
data.head()
```

```python
# PASO 3: Configuración del Experimento (Setup)
from pycaret.classification import *

# Definimos que queremos predecir la columna 'retraso'
s = setup(data, target = 'retraso', session_id = 123, verbose=False)

print("✅ Entorno de Inteligencia Artificial configurado y listo.")
```
